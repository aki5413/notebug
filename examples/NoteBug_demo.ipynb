## 1. 반복문 따라가기
`# bp` 라고 쓴 줄에는 처음부터 브레이크포인트가 걸립니다. ▶ 를 누를 때마다 `total` 과 `i` 가 어떻게 바뀌는지 보세요.

💡 멈춘 상태에서 콘솔에 `total + i` 나 `i * i` 를 입력해 보세요.

💡 콘솔에는 여러 줄도 쓸 수 있어요. `for k in range(3):` 에서 Enter → 자동 들여쓰기 → `print(total + k)` → **빈 줄에서 Enter** 로 실행. Shift+Enter 는 줄바꿈, Ctrl+Enter 는 바로 실행, ↑ 는 이전 입력.

````python
%%notebug --lang ko
total = 0
for i in range(1, 6):
    total = total + i   # bp
print('합계:', total)
````

## 2. 함수와 재귀
`breakpoint()` 도 브레이크포인트가 됩니다. **함수 안으로(F7)** 로 재귀 호출에 들어가고, 오른쪽 **호출 스택**에서 쌓이는 모습을 보세요.

````python
%%notebug --lang ko --pet dog
def fact(n):
    breakpoint()
    if n <= 1:
        return 1
    return n * fact(n - 1)

result = fact(4)
print(result)
````

## 3. 버블 정렬
안쪽 반복문의 교환(swap) 줄에 브레이크포인트를 걸고 `nums` 가 어떻게 바뀌는지 관찰하세요. 변수 패널의 ▸ 를 누르면 리스트를 펼쳐 볼 수 있어요.

````python
%%notebug --lang ko --pet hedgehog
nums = [5, 2, 9, 1, 7]
n = len(nums)
for i in range(n - 1):
    for j in range(n - 1 - i):
        if nums[j] > nums[j + 1]:
            nums[j], nums[j + 1] = nums[j + 1], nums[j]   # bp
print(nums)
````

## 4. 🔍 버그를 찾아라!
평균은 `(80 + 90 + 70) / 3 = 80` 이어야 하는데 결과가 이상합니다. 에러는 나지 않아요. 디버거로 원인을 찾아보세요.

````python
%%notebug --lang ko --pet penguin
def average(scores):
    total = 0
    for k in range(1, len(scores)):
        total += scores[k]
    return total / len(scores)

print(average([80, 90, 70]))
````

## 5. 에러가 나는 프로그램
에러가 난 줄은 빨간색으로 표시되고, 출력 창에 Traceback 이 나옵니다. ⏪ 로 한 단계씩 되돌아가 보세요.

````python
%%notebug --lang ko --pet turtle
prices = {'apple': 1200, 'banana': 800}
cart = ['apple', 'banana', 'melon']
sum_price = 0
for item in cart:
    sum_price += prices[item]
print(sum_price)
````

## 6. 같은 리스트? 다른 리스트?
변수 옆 보라색 **#번호**가 같으면 같은 객체입니다. 마우스를 올려 보세요. `b.append(4)` 뒤에 `a` 는 어떻게 될까요?

````python
%%notebug --lang ko --pet dino
a = [1, 2, 3]
b = a
c = a[:]
b.append(4)
print(a, b, c)
````

## 8. 데이터 분석 (numpy · pandas · matplotlib)
DataFrame 은 변수 창에서 ▸ 를 누르면 표로 볼 수 있고, 그래프는 **출력 창**에 그 단계에서 나타납니다 (클릭하면 크게). `apply` 에 넘긴 lambda 는 한 단계로 처리되어 1000 행이라도 금방 끝나요.

💡 9행에서 멈춘 뒤 콘솔에 `df[df.grade == 'fail'].shape` 나 `plt.plot(df.score.values[:20], 'o-')` 를 입력해 보세요.

````python
%%notebug --lang ko
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)
df = pd.DataFrame({'name': ['A', 'B', 'C', 'D'] * 250,
                   'score': np.random.randint(50, 100, 1000)})
df['grade'] = df['score'].apply(lambda s: 'pass' if s >= 60 else 'fail')
mean = df.groupby('name')['score'].mean()   # bp
print(mean.round(1))
plt.bar(mean.index, mean.values)
plt.title('mean score')
plt.show()
df.head()
````

## 옵션 / オプション
```
%%notebug --lang ko|ja|en  --theme night|light|darcula|contrast  --font 16  --pet cat|dog|...  --power  --sound  --max 3000  --no-achievements  --no-export
```
- `--theme` 테마 (light 는 프로젝터용, darcula 는 PyCharm 풍, contrast 는 고대비), `--font` 글자 크기(px) — 화면 전체가 같이 커집니다
- `--power` 파워 모드(콤보·이펙트), `--sound` 효과음 (`--no-power` 로 강제 끄기)
- `--max` 기록할 최대 단계 수 (무한 반복 방지)
- `--no-achievements` 🏆 업적 알림 끄기
- 디버거로 실행한 셀의 변수는 보통 셀처럼 다음 셀에서도 쓸 수 있습니다 (`--no-export` 로 끄기)
- 콘솔(>>>)은 Google Colab 에서 동작합니다.

노트 전체 기본값은 `notebug.config(lang='ko', theme='light', font_size=16, achievements=False)` 처럼 한 번에 정할 수 있어요. 색을 하나씩 바꾸려면 `notebug.config(colors={'cur': 'rgba(0,160,255,.25)', 'kw': '#0000cc'})`.

````python
%%notebug --lang ja --theme light --font 16 --power --sound
words = ['apple', 'kiwi', 'banana']
lengths = {}
for w in words:
    lengths[w] = len(w)
print(lengths)
````

## 모듈로 쓰기 / モジュールとして使う

````python
code = """
scores = [70, 85, 90]
best = max(scores)   # bp
print(best)
"""
notebug.run(code, pet='penguin')              # 문자열 코드 디버깅

open('hw1.py', 'w').write(code)
notebug.run_file('hw1.py')                     # .py 파일 디버깅
notebug.save_html(code, 'hw1_debug.html')      # 브라우저용 HTML 로 저장
````
